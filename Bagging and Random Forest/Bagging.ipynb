## Ensembles

**Ансамбли (ensembles)** - набор элементов, которые в совокупности вносят свой вклад в целое.

Хорошим примером ансамблей считается теорема Кондорсе «о жюри присяжных» (1784). Если каждый член жюри присяжных имеет независимое мнение, и если вероятность правильного решения члена жюри больше 0.5, то тогда вероятность правильного решения присяжных в целом возрастает с увеличением количества членов жюри и стремится к единице. Если же вероятность быть правым у каждого из членов жюри меньше 0.5, то вероятность принятия правильного решения присяжными в целом монотонно уменьшается и стремится к нулю с увеличением количества присяжных.

### Bootstarapping

Bagging, от bootstrap aggregation - придуман в 1994 году. Основан на ститстическом методе бутстрепа, который позволяет оценивать многоин статистики сложных распредлелений.

Пусть имеется выбока $X$ размера $N$. Равномерно возьмем из выборки $N$ объектов с возвращением. Повторяя процедуру $M$ раз, сгенерируем M подвыборок $X_1,...,X_M$. Теперь мы имеем достаточно большое число выборок и можем оценивать различные статистики исходного распределения.

### Bagging

Пусть имеется обучающая выборка $\large X$. С помощью бутстрэпа сгенерируем из неё выборки $\large X_1, \dots, X_M$. Теперь на каждой выборке обучим свой классификатор $\large a_i(x)$. Итоговый классификатор будет усреднять ответы всех этих алгоритмов (в случае классификации это соответствует голосованию): $\large a(x) = \frac{1}{M}\sum_{i = 1}^M a_i(x)$. Рассмотрим задачу регрессии с базовыми алгоритмами $b_1(x), \dots , b_n(x)$. Предположим, что существует истинная функция ответа для всех объектов $y(x)$, а также задано распределение на объектах $p(x)$. В этом случае мы можем записать ошибку каждой функции регрессии. $\varepsilon_i(x) = b_i(x) - y(x), i = 1, \dots, n$

Матожидание среднеквадратичной ошибки: $\mathbb{E}_x\left[\left(b_i(x) - y(x)\right)^{2}\right] = \mathbb{E}_x\left[\varepsilon_i^{2}(x)\right]$

Средняя ошибка построенных функций регрессии имеет вид: $\mathbb{E}_1 = \frac{1}{n}\mathbb{E}_x \sum_{i=1}^n \varepsilon_i^{2}(x)$

Предположим, что ошибки несмещены и некоррелированы:
$\begin{array}{rcl} \mathbb{E}_x\varepsilon_i(x) &=& 0, \\ \mathbb{E}_x\varepsilon_i(x)\varepsilon_j(x) &=& 0, i \neq j. \end{array}$

Построим теперь новую функцию регрессии, которая будет усреднять ответы построенных нами функций: $a(x) = \frac{1}{n}\sum_{i=1}^{n}b_i(x)$

Найдем ее среднеквадратичную ошибку:

$\begin{array}{rcl}\mathbb{E}_n &=& \mathbb{E}_x\Big(\frac{1}{n}\sum_{i=1}^{n}b_i(x)-y(x)\Big)^2 \\ &=& \mathbb{E}_x\Big(\frac{1}{n}\sum_{i=1}^{n}\varepsilon_i\Big)^2 \\ &=& \frac{1}{n^2}\mathbb{E}_x\Big(\sum_{i=1}^{n}\varepsilon_i^2(x) + \sum_{i \neq j}\varepsilon_i(x)\varepsilon_j(x)\Big) \\ &=& \frac{1}{n}\mathbb{E}_1\end{array}$

Таким образом, усреднение ответов позволило уменьшить средний квадрат ошибки в n раз!

Бэгинг позволяет снизить дисперсию (variance), уменьшая величину, на сколько ошибка будет отличаться, если обучать модель на разных наборах, то есть предотвращает переобучение.

В `scikit-learn` есть реализации `BaggingRegressor` и `BaggingClassifier`, которые позволяют использовать большинство алгоритмов "внутри". Рассмотрим на практике, как работает бэггинг, и сравшим с деревом решений.

Бэггинг эффективен на малых выборках, когда исключение даже малой части обучающих объектов приводит к построению существенно различных базовых классификаторов.

Следует отметить, что рассмотренный нами пример не очень применим на практике, поскольку мы сделали предположение о некоррелированности ошибок, что редко выполняется. Если это предположение неверно, то уменьшение ошибки оказывается не таким значительным.

### Out-of-bag error

При использовании случайных лесов нет необходимости в кросс-валидации или в отдельно валидационном сете.

Каждое дерево строится с использованием разных образцов бутстрэпа из исходных данных. Примерно 37% примеров остаются вне выборки бутстрэпа и не используются при построении k-го дерева.

Это можно легко доказать: пусть в выборке $\ell$ объектов. На каждом шаге все объекты попадают в подвыборку с возвращением равновероятно, т.е отдельный объект — с вероятностью $\large\frac{1}{\ell}.$ Вероятность того, что объект НЕ попадет в подвыборку (т.е. его не взяли $\ell$ раз): $(1 - \frac{1}{\ell})^\ell$. При $\ell \rightarrow +\infty$, получаем второй замечательный предел. Тогда вероятность попадания объекта в подвыборку $\approx 1 - \frac{1}{e} \approx 63\%$.

Получается, что каждый базовый алгоритм обучается на ~63% исходных объектов. Значит, на оставшихся ~37% его можно сразу проверять. Out-of-Bag оценка — это усредненная оценка базовых алгоритмов на тех ~37% данных, на которых они не обучались.
