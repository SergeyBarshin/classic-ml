### Feature importance

В RF можно довольно просто получить информацию о том какие переменные больше всего влияют на результат.

Если построить много деревьев решений (случайный лес), то чем выше в среднем признак в дереве решений, тем он важнее в данной задаче классификации/регрессии.


Чем больше уменьшается точность предсказаний из-за исключения (или перестановки) одной переменной, тем важнее эта переменная, и поэтому переменные с бо́льшим средним уменьшением точности более важны для классификации данных. Каждый раз, когда отдельная переменная используется для разбиения узла, неопределенность Джини для дочерних узлов рассчитывается и сравнивается с коэффициентом исходного узла. Изменения в значении критерия разделения суммируются для каждой переменной и нормируются в конце вычисления. Переменные, которые приводят к узлам с более высокой чистотой, имеют более высокое снижение коэффициента Джини.

На рисунке выше видно, что люди больше всего обращают внимание на персонал и соотношение цена/качество и на основе впечатления от данных вещей пишут свои отзывы. Но разница между этими признаками и менее влиятельными признаками не очень значительная, и выкидывание какого-то признака приведет к уменьшению точности нашей модели. Но даже на основе нашего анализа мы можем дать рекомендации отелям в первую очередь лучше готовить персонал и/или улучшить качество до заявленной цены.

## Плюсы и минусы случайного леса
Плюсы:
1. высокая точность предсказания, в большинстве задач лучше лин. моделей. Сравнима с бустингом/
2. не чувствителен к выбросам из-зи случайного семплирования.
3. не чувствителен к мастабированиею
4. не требует тщательного тюнинга, он позволяет достичь всего 0.5-3% прироста.
5. способен эффективно обрабатывать данные с большим числом признаков и классов
6. хорошо работает и непрерывными и с дискретными величинами
7. методы оценивания значимости отдельных признаков в модели
8. хорошо работает с пропустками
9. вычисляет близость между парами объектов, которые могут использоваться при кластеризации, обнаружении выбросов или (путем масштабирования) дают интересные представления данных

Минусы:
1. в отличие от одного дерева, результаты случайного леса сложнее интерпретировать
2. нет формальных выводов (p-values), доступных для оценки важности переменных
3. алгоритм работает хуже многих линейных методов, когда в выборке очень много разреженных признаков (тексты, Bag of words)
4. случайный лес не умеет экстраполировать, в отличие от той же линейной регрессии (но это можно считать и плюсом, так как не будет экстремальных значений в случае попадания выброса)
5. алгоритм склонен к переобучению на некоторых задачах, особенно на зашумленных данных
6. для данных, включающих категориальные переменные с различным количеством уровней, случайные леса предвзяты в пользу признаков с большим количеством уровней: когда у признака много уровней, дерево будет сильнее подстраиваться именно под эти признаки, так как на них можно получить более высокое значение оптимизируемого функционала (типа прироста информации)
7. больший размер получающихся моделей. Требуется $\large O(NK) $ памяти для хранения модели, где $\large K $ — число деревьев.
